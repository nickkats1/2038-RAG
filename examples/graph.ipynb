```python
from pathlib import Path

from dotenv import load_dotenv
from langchain_openai import ChatOpenAI

from src.data_utils import load_documents, chunker
from src.graph.retriever import GraphRetriever

project_dir = Path("/home/nick/new-speedrun")

load_dotenv(project_dir / ".env")

llm = ChatOpenAI(model="gpt-4o", temperature=0.0)
```

```python
documents = load_documents(
    file_path=str(project_dir / "data" / "2038.pdf")
)

chunks = chunker(
    documents=documents,
    chunk_size=300,
    chunk_overlap=30,
)

print("Pages:", len(documents))
print("Chunks:", len(chunks))
print("\nFirst chunk:\n", chunks[0].page_content)
```

```python
graph_retriever = GraphRetriever(
    llm=llm,
    hops=3,
)

sample_chunks = chunks[10:80]

graph_retriever.add_documents(sample_chunks)

print("Indexed chunks:", len(graph_retriever.documents))
print("Entities:", graph_retriever.graph.number_of_nodes())
print("Relationships:", graph_retriever.graph.number_of_edges())
```

```python
entities = list(graph_retriever.graph.nodes)

print("Entities:")
for entity in entities:
    print(entity)

print("\nRelationships:")
for subject, obj, data in graph_retriever.graph.edges(data=True):
    print(f"{subject} — {data['relation']} → {obj}")
```
